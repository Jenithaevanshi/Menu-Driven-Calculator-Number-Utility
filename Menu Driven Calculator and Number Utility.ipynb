```python
def get_number(message):
    return float(input(message))


def get_integer(message):
    return int(input(message))


# ---------------- Calculator Functions ----------------

def addition():
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")
    print("Addition =", num1 + num2)


def subtraction():
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")
    print("Subtraction =", num1 - num2)


def multiplication():
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")
    print("Multiplication =", num1 * num2)


def division():
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")

    if num2 == 0:
        print("Division by zero is not allowed.")
    else:
        print("Division =", num1 / num2)


def percentage():
    obtained = get_number("Enter the obtained marks: ")
    total = get_number("Enter the total marks: ")

    if total == 0:
        print("Total marks cannot be zero.")
    else:
        percentage = (obtained / total) * 100
        print("Percentage =", round(percentage, 2), "%")


def even_odd():
    number = get_integer("Enter an integer: ")

    if number % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")


def compare_numbers():
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")

    if num1 > num2:
        print(num1, "is larger")
    elif num2 > num1:
        print(num2, "is larger")
    else:
        print("Both numbers are equal")



# ---------------- Main Program ----------------

while True:

    print("\n====== Menu Driven Calculator & Number Utility ======")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Percentage")
    print("6. Even / Odd")
    print("7. Compare Numbers")
    print("8. Exit")

    choice = input("Enter your choice (1-8): ")

    match choice:

        case "1":
            addition()

        case "2":
            subtraction()

        case "3":
            multiplication()

        case "4":
            division()

        case "5":
            percentage()

        case "6":
            even_odd()

        case "7":
            compare_numbers()

        case "8":
            print("Thank you!")
            break

        case _:
            print("Invalid choice! Please enter a number between 1 and 8.")
```
